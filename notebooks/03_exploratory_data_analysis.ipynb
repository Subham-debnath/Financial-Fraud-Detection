Step 1: Import Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

plt.style.use("ggplot")
```

Step 2: Load Dataset

```python
df = pd.read_csv("../data/processed/creditcard_clean.csv")
```

Step 3: Display First Rows

```python
df.head()
```

Step 4: Dataset Information

```python
df.info()
```

Step 5: Dataset Shape

```python
print(df.shape)
```

Step 6: Statistical Summary

```python
df.describe()
```

Step 7: Class Distribution

```python
df["Class"].value_counts()
```

```python
(df["Class"].value_counts(normalize=True))*100
```

Step 8: Visualize Class Distribution

```python
plt.figure(figsize=(6,5))

sns.countplot(x="Class", data=df)

plt.title("Fraud vs Legitimate Transactions")

plt.show()
```

Step 9: Pie Chart

```python
df["Class"].value_counts().plot(
    kind="pie",
    labels=["Legitimate","Fraud"],
    autopct="%1.2f%%"
)

plt.title("Transaction Distribution")

plt.ylabel("")

plt.show()
```

Step 10: Transaction Amount Distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Amount"], bins=60)

plt.title("Transaction Amount Distribution")

plt.xlabel("Amount")

plt.savefig(
    "../figures/amount_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Step 11: Transaction Time Distribution

```python
plt.figure(figsize=(10,5))

sns.histplot(df["Time"], bins=60)

plt.title("Transaction Time Distribution")

plt.xlabel("Time")

plt.savefig(
    "../figures/time_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Step 12: Boxplot of Amount

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df["Amount"])

plt.title("Boxplot of Transaction Amount")

plt.show()
```

Step 13: Boxplot of Time

```python
plt.figure(figsize=(8,4))

sns.boxplot(x=df["Time"])

plt.title("Boxplot of Time")

plt.show()
```

Step 14: Fraud vs Amount

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x="Class",
    y="Amount",
    data=df
)

plt.title("Fraud vs Transaction Amount")

plt.savefig(
    "../figures/fraud_vs_amount.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Step 15: Fraud vs Time

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x="Class",
    y="Time",
    data=df
)

plt.title("Fraud vs Transaction Time")

plt.savefig(
    "../figures/fraud_vs_time.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Step 16: Correlation Matrix

```python
corr = df.corr()

plt.figure(figsize=(18,14))

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix")

plt.savefig(
    "../figures/correlation_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

Step 17: Correlation with Target

```python
corr = df.corr()["Class"].sort_values(
    ascending=False
)

print(corr)
```

Step 18: Top Correlated Features

```python
top_features = corr.abs().sort_values(
    ascending=False
).head(10)

print(top_features)
```

Step 19: Pairplot (Selected Features)

```python
sample = df.sample(
    3000,
    random_state=42
)
```

```python
sns.pairplot(
    sample[
        [
            "V10",
            "V12",
            "V14",
            "Amount",
            "Class"
        ]
    ],
    hue="Class"
)

plt.show()
```
