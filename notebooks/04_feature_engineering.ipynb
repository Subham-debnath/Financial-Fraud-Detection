Step 1: Import Libraries

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier

%matplotlib inline
```

Step 2: Load Dataset

```python
df = pd.read_csv("../data/processed/creditcard_clean.csv")
```

Step 3: Display Dataset

```python
df.head()
```

Step 4: Separate Features and Target

```python
X = df.drop("Class", axis=1)

y = df["Class"]
```

Step 5: Verify Features

```python
print(X.columns)
```

Step 6: Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

Step 7: Feature Scaling

```python
scaler = StandardScaler()

X_train[["Time","Amount"]] = scaler.fit_transform(
    X_train[["Time","Amount"]]
)

X_test[["Time","Amount"]] = scaler.transform(
    X_test[["Time","Amount"]]
)
```

Step 8: Verify Scaling

```python
X_train[["Time","Amount"]].describe()
```

Step 9: check number of features

```python
print("Number of Features:", X_train.shape[1])

print("\nFeature Names:\n")

print(X_train.columns.tolist())
```

Step 10: Identify Features That Need Scaling

```python
X_train[["Time", "Amount"]].describe()
```

Step 11: Check for Constant Columns

```python
constant_columns = []

for col in X_train.columns:
    if X_train[col].nunique() == 1:
        constant_columns.append(col)

print("Constant Columns:")
print(constant_columns)
```

Step 12: Verify Final Feature Matrix

```python
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

Step 13: Save the Engineered Dataset

```python
X_train.to_csv(
    "../data/processed/X_train_engineered.csv",
    index=False
)

X_test.to_csv(
    "../data/processed/X_test_engineered.csv",
    index=False
)

y_train.to_csv(
    "../data/processed/y_train.csv",
    index=False
)

y_test.to_csv(
    "../data/processed/y_test.csv",
    index=False
)
```
